# Custom Tools

```python
from langchain_community.tools import tool

```

### step 1: make a function

```python
def multiply(a , b):
    # this function is used for multiplication
    return a*b
```

### step 2 : add type hints

```python
def multiply(a : int , b : int) -> int:
    # this function is used for multiplication
    return a*b
```

### step 3: Add tool decorator

```python
@tool
def multiply(a : int , b : int) -> int:
    """this function is used for multiplication"""
    return a*b
```

```python
result = multiply.invoke({"a":5, "b":4})
```

```python
print(result)
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

# Method 2 : using structuredTools

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True , description="the first number for multiplication")
    b: int = Field(required=True, description = "the second number for multiplication")
    
```

```python
def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_func,
    name="multiply",
    description = "multiplication of two numbers but input are validated in pydantic",
    args_schema = MultiplyInput
    
)
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Method 3 - Using BaseTool Class

```python
from langchain_community.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to multiply")
    b: int = Field(required=True, description="The second number to multiply")
```

```python
class multiply_tool(BaseTool):
    name : str ="multiplication"
    description :str = "Mulitplication of two numbers "
    args_schema : type(BaseModel) = MultiplyInput
    
    def _run(self, a:int , b:int)->int:
        return a*b 
```

```python
multiply_tool = multiply_tool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

# ToolKit

```python
from langchain_core.tools import tool


# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
